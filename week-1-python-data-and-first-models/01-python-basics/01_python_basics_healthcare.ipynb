## 1. Python Basic Data Types

Python provides different data types to represent
different kinds of information.

Here we demonstrate integer, float, string and
boolean data types using healthcare-related examples.

```python
patient_age = 45

body_temperature = 37.5

patient_category = "Adult"

has_diabetes = True

print("Patient Age:", patient_age)
print("Body Temperature:", body_temperature)
print("Patient Category:", patient_category)
print("Has Diabetes:", has_diabetes)

print("\nData Types:")

print(type(patient_age))
print(type(body_temperature))
print(type(patient_category))
print(type(has_diabetes))
```

## 2. Python Lists

A list is used to store multiple items in a single variable.
Here, we create a list containing five healthcare-related
departments and access the third item.

```python
departments = ["Cardiology", "Neurology", "Orthopedics", "Dermatology", "Pediatrics"]

print("Healthcare Departments:", departments)

print("Third Department:", departments[2])
```

## 3. Python Dictionaries

A dictionary stores information as key-value pairs.
Here, we create a sample patient record containing
basic healthcare-related information.

```python
patient = {"patient_id": 101, "age": 45, "blood_group": "O+", "body_temperature": 37.5, "has_diabetes": True}

print("Patient Record:", patient)

print("Patient Age:", patient["age"])
print("Blood Group:", patient["blood_group"])
```

## 4. Python For Loops

A for loop is used to iterate over a sequence
such as a list.

Here, we iterate through the healthcare departments
and print each department individually.

```python
for department in departments:
    print(department)
```

## 5. Conditional Statements

Conditional statements allow a program to make decisions
based on whether a condition is True or False.

Here, we classify a patient's body temperature into
different categories using if, elif and else.

```python
# Sample body temperature
body_temperature = 38.5

# Classify the temperature
if body_temperature >= 38:
    print("High temperature")
elif body_temperature >= 37:
    print("Slightly elevated temperature")
else:
    print("Normal temperature")
```

## 6. Python Functions

A function is a reusable block of code that performs
a specific operation.

Here, we define a function that accepts two arguments
and returns their sum.

```python
def calculate_total(a, b):
    result = a + b
    return result
```

### Calling the Function

We call the previously defined function with two
arguments and store its returned value in a variable.

```python
total = calculate_total(100, 200)
print("Total:", total)

final_total = total * 2
print("Final Total:", final_total)
```

#Another Example

```python
def calculate_total(a, b):
    return a + b

total = calculate_total(100, 200)

discounted_total = total - 50
print(discounted_total)
```

## 7. GPU Configuration and Verification

A GPU accelerates computationally intensive operations,
especially machine learning and deep learning workloads.

Here, we use PyTorch to verify GPU availability
in the Google Colab environment.

```python
import torch

# Check GPU availability
print("GPU Available:", torch.cuda.is_available())

# Display GPU name if available
if torch.cuda.is_available():
    print("GPU Name:", torch.cuda.get_device_name(0))
else:
    print("No GPU detected.")
```
