```python
marks = 85

if marks >= 90:
    print("Grade: A")

elif marks >= 75:
    print("Grade: B")

elif marks >= 50:
    print("Grade: C")

else:
    print("Grade: Fail")
```

```python
def calculate_average(mark1, mark2):
    average = (mark1 + mark2) / 2
    return average

result = calculate_average(85, 90)

print("Average:", result)
```
