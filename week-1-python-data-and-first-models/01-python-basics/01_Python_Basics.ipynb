## Task 1 – Variables and basic data types
Create one variable of each required type (`int`, `float`, `str`, `bool`) and display its value and type.

```python
login_attempts = 5                  # int
risk_score = 0.82                    # float
clinic_name = "Green Valley Clinic"  # str
account_locked = False               # bool

for variable_name, value in {
    "login_attempts": login_attempts,
    "risk_score": risk_score,
    "clinic_name": clinic_name,
    "account_locked": account_locked
}.items():
    print(variable_name, "=", value, "| type:", type(value).__name__)
```

## Task 2 – Lists
Create a list with at least five items and print the third item. Python list indexing starts at zero.

```python
security_events = [
    "Successful login",
    "Failed login",
    "Password reset",
    "Multiple failed attempts",
    "Unusual login location"
]
print("Security events:", security_events)
print("Third item:", security_events[2])
```

## Task 3 – Dictionary
Create a dictionary describing a realistic clinic security event. It contains more than four key-value pairs.

```python
login_record = {
    "user_id": "CLN104",
    "role": "Receptionist",
    "login_time": "09:15",
    "failed_attempts": 3,
    "source": "Clinic workstation",
    "status": "Under review"
}
for key, value in login_record.items():
    print(f"{key}: {value}")
```

## Task 4 – For loop
Iterate over the list and display each event with its sequence number.

```python
for number, event in enumerate(security_events, start=1):
    print(f"{number}. {event}")
```

## Task 5 – if / elif / else
Use a conditional statement to categorise login risk based on the number of failed attempts.

```python
failed_attempts = 3

if failed_attempts >= 5:
    risk_level = "High"
elif failed_attempts >= 2:
    risk_level = "Medium"
else:
    risk_level = "Low"

print("Failed attempts:", failed_attempts)
print("Risk level:", risk_level)
```

## Task 6 – Function with two arguments and a return value
The function must return a value, not merely print it. The returned value is stored and reused in the next cell.

```python
def calculate_total_attempts(successful_attempts, failed_attempts):
    """Return the total number of login attempts."""
    return successful_attempts + failed_attempts

total_attempts = calculate_total_attempts(12, 3)
print("Returned total:", total_attempts)
```

## Reuse the returned value
Use the value returned by the function in another calculation.

```python
attempts_after_review = total_attempts * 2
print("Reused returned value:", attempts_after_review)
```
