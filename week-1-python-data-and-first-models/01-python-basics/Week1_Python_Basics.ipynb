# 1. Python Basic Data Types

This section demonstrates the four basic Python data types:
integer, float, string and boolean.

```python

age = 20
temperature = 28.5
name = "Amirthavalli"
is_student = True
print("Integer:", age)
print("Float:", temperature)
print("String:", name)
print("Boolean:", is_student)
print("\nData Types:")
print(type(age))
print(type(temperature))
print(type(name))
print(type(is_student))
```

# 2. Python Lists

A list stores multiple items in a single variable.
Here, we create a list of five crops and print the third item.

```python
crops = ["Rice", "Wheat", "Maize", "Cotton", "Sugarcane"]

print("List of crops:", crops)
print("Third crop:", crops[2])
```

# 3. Python Dictionary

A dictionary stores information in key-value pairs.
Here, we create a dictionary containing details about a rice crop.


```python
crop_details = {
    "crop_name": "Rice",
    "soil_type": "Clay Soil",
    "water_requirement": "High",
    "season": "Kharif"
}

print("Crop Details:")

for key, value in crop_details.items():
    print(key, ":", value)
```

# 4. Python For Loop

A for loop is used to repeat a block of code.
Here, we use a for loop to display each crop in the list.


```python
crops = ["Rice", "Wheat", "Maize", "Cotton", "Sugarcane"]

print("Available Crops:")

for crop in crops:
    print(crop)
```

# 5. Conditional Statements

Conditional statements help a program make decisions.
Here, we check the soil moisture level and display a suitable message.

```python
soil_moisture = 25

if soil_moisture < 20:
    print("Soil is dry. Watering is required.")

elif soil_moisture <= 40:
    print("Soil moisture is moderate.")

else:
    print("Soil moisture is sufficient.")
```

# 6. Python Functions

A function is a reusable block of code.
Here, we create a function that accepts two arguments and returns their sum.

```python
def calculate_total(a, b):
    return a + b

result = calculate_total(100, 200)

print("Total:", result)
```

# 7. Difference Between Print and Return

The print() statement displays a value on the screen.
The return statement sends a value back from a function so it can be reused.

```python
def add_wrong(a, b):
    print(a + b)

def add_right(a, b):
    return a + b

total = add_right(2, 3) * 10

print("Correct result:", total)

broken = add_wrong(2, 3)

print("Returned value:", broken)
```

# 8. Checking GPU Availability

Google Colab provides GPU support for machine learning tasks.
