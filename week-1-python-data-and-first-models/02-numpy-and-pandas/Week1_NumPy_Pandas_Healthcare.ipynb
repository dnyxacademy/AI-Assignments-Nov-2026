## Step 1: Import Libraries and Load Dataset

In this step, we import Pandas and NumPy and load the healthcare dataset into a DataFrame.


```python
# Install the UCI dataset package
!pip install -q ucimlrepo

import pandas as pd
import numpy as np
from ucimlrepo import fetch_ucirepo

# Download the CDC Diabetes Health Indicators dataset
dataset = fetch_ucirepo(id=891)

# Combine features and target into one DataFrame
df = pd.concat(
    [dataset.data.features, dataset.data.targets],
    axis=1
)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())
```

## Step 7: Data Selection and Filtering

We select individual columns, multiple columns and filter healthcare records based on age.

```python
# Select one column
display(df["Age"].head())

# Select two columns
display(df[["Age", "BMI"]].head())

# Filter people aged above 50
filtered_data = df[df["Age"] > 50]

print("Records with age category above 50:")
display(filtered_data.head())
print("Filtered records:", len(filtered_data))
```

## Step 8: NumPy Performance Comparison

We compare the time taken to calculate the sum of BMI values using a Python loop and NumPy vectorisation.

```python
values = df["BMI"].to_numpy()

# Python loop
def loop_sum():
    total = 0
    for value in values:
        total += value
    return total

# NumPy vectorised sum
def numpy_sum():
    return values.sum()

print("Loop sum:", loop_sum())
print("NumPy sum:", numpy_sum())

%timeit loop_sum()
%timeit numpy_sum()
```
