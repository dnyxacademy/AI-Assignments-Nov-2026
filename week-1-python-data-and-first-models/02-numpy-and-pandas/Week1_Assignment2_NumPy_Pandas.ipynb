```python
import numpy as np

# Convert final grades into NumPy array
values = df["G3"].to_numpy()

# Method 1: Traditional Python loop
%timeit sum(v for v in values)

# Method 2: NumPy vectorisation
%timeit values.sum()

# Verify both produce the same result
print("Loop Sum:", sum(v for v in values))
print("NumPy Sum:", values.sum())
```
