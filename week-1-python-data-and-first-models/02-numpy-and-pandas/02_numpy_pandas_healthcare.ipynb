##NumPy: Loop vs Vectorised Operations
We use NumPy to calculate the sum of the Age column using a Python loop and a vectorised operation. We then compare their execution times using %timeit.

```python
import numpy as np

values = df['Age'].to_numpy()

print("Number of Values:", len(values))
print("First 5 Values:", values[:5])
```

##Python Loop Summation:
We iterate through every age value and add it to a running total.

```python
loop_sum = 0

for value in values:
    loop_sum += value

print("Sum using Python Loop:", loop_sum)
```

Sum using NumPy vectorisation

```python
numpy_sum = values.sum()

print("Sum using NumPy:", numpy_sum)
```

Execution Time Analysis:

```python
%timeit sum(value for value in values)
```

```python
%timeit values.sum()
```

### NumPy Performance Comparison – Observations

The sum of the Age column was 4,099,322 using both approaches.
The Python loop took an average of 12.7 ms per execution.
The NumPy vectorised operation took an average of 24.5 microseconds per execution.
Based on these measurements, NumPy was approximately 518.4 times faster than the Python loop.
This demonstrates the performance advantage of NumPy vectorised operations when processing large numerical arrays.
